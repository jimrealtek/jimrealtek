```python
#Extracing independent variable
x=dataset.iloc[:,:-1].values
x
```

```python
#Extracting the depent variable
y = dataset.iloc[:,-1].values
y
```

```python
#Handling missing data
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=nm.nan, strategy='mean')
imputer.fit(x[:,2:3])
x[:,2:3] = imputer.transform(x[:,2:3])
x
```

```python
#Encoding categorical data
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
labelencoder_x = LabelEncoder()
x[:,0] = labelencoder_x.fit_transform(x[:,0])
x
#
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
import numpy as nm

# First, handle missing values in Department so OneHotEncoder doesn't fail
# Assuming x contains ['S/N', 'Name', 'Age', 'Department']

# Apply OneHotEncoder specifically to column index 3 (Department)
ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(drop='first', handle_unknown='ignore'), [3])],
    remainder='passthrough'
)

# Transform x
x_encoded = ct.fit_transform(x)

print(x_encoded)
```

```python
#For purchased variable
labelencoder_y = LabelEncoder()
y = labelencoder_y.fit_transform(y)
y
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
import numpy as np

# Assuming x contains ['S/N', 'Name', 'Age', 'Department']
x_features = x[:, 2:4]

ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(drop='first'), [1])],
    remainder='passthrough'
)
x_encoded = ct.fit_transform(x_features)

x_train, x_test, y_train, y_test = train_test_split(x_encoded, y, test_size=0.2, random_state=1)
```
