**Extracting dependent and independent variable**

In machine learning, it is important to distinguish the matrix of features(independent variables) and dependent variables from the dataset, there are three independent variables that are country, age, and salary, and one is dependent variable which is purchased.

```python
#Extracting independent variable
'''To extract the independent variable, we use iloc[] method of pandas library.It is used to extract the required rows and columns from the dataset.'''

x=dataset.iloc[:,:-1].values
x
```

**Ways of handling missing data**

The next step of data preprocessing is to handle missing data in the dataset. If our dataset contains some missing data, then it may create a huge problem for our machine learning model. Hence, it is necessary ot handle missing values present in the datasets.

There are mainly two ways to handle missing data:

1. By deleting the particular row (Analytical Model)

2. By Calculating the mean to fill the missing value(Learned Model)

```python
#Handling missing data
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(missing_values=nm.nan,strategy='mean')
imputer=imputer.fit(x[:,1:3])
x[:,1:3]=imputer.transform(x[:,1:3])
x
```

**Encoding categorical data**

Categorical data is data which has same categories in our dataset, there are two categorical variable which are country and purchased. Since machine learning model completely works on mathematics and numbers, but in our dataset would have a categorical variable, then it may create trouble while building the model. So it is necessary to encode these categorical variables in to numbers.

```python
#Encoding categorical data
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
labelencoder_x=LabelEncoder()
x[:,0]=labelencoder_x.fit_transform(x[:,0])
x
```

**Dummy Variable**

Dummy variables are those variables which have values 0 or 1. The 1 value give the presence of that variables in a particular column and rest variablle become 0. With dummy encoding, we will have a number of columns eqaul to the number of categorinies. So it will produce three columns having 0 and 1 values.

```python
#Dummy Variable
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder # Import OneHotEncoder

columnTransformer=ColumnTransformer([('encoder',OneHotEncoder(),[0])],
                                      remainder='passthrough')
x=np.array(columnTransformer.fit_transform(x))
x
```
